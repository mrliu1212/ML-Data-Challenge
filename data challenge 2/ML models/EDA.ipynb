```python
import pandas as pd
train_data = pd.read_csv("../data/train.csv")
test_data = pd.read_csv("../data/test.csv")

train_data = train_data.drop("w",axis=1)
train_data.info()
```

```python
# deal with categorical columns
categorical_feature = []
for item in train_data.columns:
    if train_data[item].dtype == object:
        categorical_feature.append(item)
print(f'Categorical columns are {categorical_feature}')
train_data_encoded = pd.get_dummies(train_data, columns=categorical_feature)
test_data_encoded = pd.get_dummies(test_data, columns=categorical_feature)
# train_data_encoded.info()

# check null value
print(f'Is there any null value in the train dataset? {train_data_encoded.isnull().any(axis=1).any(axis=0)}')
print(f'Is there any null value in the test dataset? {train_data_encoded.isnull().any(axis=1).any(axis=0)}')

# I am not going to check multicollinearity because we are going to use model that are resistent to it (Elastic Net, RF, GBM)
```

```python
# save new processed file
train_data_encoded.to_csv("../data/train_encoded.csv",index=False)
test_data_encoded.to_csv("../data/test_encoded.csv",index=False)
```
